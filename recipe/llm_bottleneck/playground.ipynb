````python
# Alternative: Direct inference using VERL framework (if vLLM fails)

# This approach loads the model using the VERL framework directly
# which can handle FSDP checkpoints without conversion

try:
    from verl.workers.engine.fsdp.engine_impl import FSDPEngine
    from verl.utils.config import omega_conf_to_dataclass
    from omegaconf import OmegaConf
    import torch
    
    # Create a minimal config for the model
    config_dict = {
        "model": {
            "path": "/projectnb/noc-lab/ylchen/llm_bottleneck/checkpoints/llm-bottleneck/bs_128_length_4096_full_training_distill_1.5b/global_step_400/actor",
            "trust_remote_code": False
        },
        "checkpoint": {
            "load": ["model"],
            "save": []
        }
    }
    
    config = OmegaConf.create(config_dict)
    
    # Initialize the FSDP engine
    engine = FSDPEngine(config)
    engine.init_model()
    
    # Load the checkpoint
    engine.load_checkpoint("/projectnb/noc-lab/ylchen/llm_bottleneck/checkpoints/llm-bottleneck/bs_128_length_4096_full_training_distill_1.5b/global_step_400/actor")
    
    print("Successfully loaded model using VERL FSDP engine")
    print("Note: This approach requires distributed training setup and may not work in a single-GPU environment")
    
except Exception as e:
    print(f"VERL framework approach failed: {e}")
    print("This is expected in a single-GPU environment. Use the HuggingFace conversion approach instead.")
````
